### 1. ¿Qué diferencias encuentras entre las estadísticas originales y las del escalado MinMax?


```python
from sklearn.datasets import load_iris
from sklearn.preprocessing import MinMaxScaler, StandardScaler
import pandas as pd

# Cargar y escalar datos
iris = load_iris(as_frame=True)
data = iris.data

minmax_scaler = MinMaxScaler()
data_minmax = pd.DataFrame(
    minmax_scaler.fit_transform(data), columns=data.columns
)

standard_scaler = StandardScaler()
data_standard = pd.DataFrame(
    standard_scaler.fit_transform(data), columns=data.columns
)

# Imprimir estadísticas
print("--- Estadísticas Originales ---")
display(data.describe())

print("--- Estadísticas MinMax ---")
display(data_minmax.describe())
```

### 2. Crea una gráfica de la distribución de la variable 'sepal length (cm)' antes y después del escalado.

```python
import matplotlib.pyplot as plt
import seaborn as sns

fig, axes = plt.subplots(1, 3, figsize=(15, 4))

# Original
sns.histplot(
    data['sepal length (cm)'],
    kde=True,
    ax=axes[0],
    color='gray',
    alpha=0.5
)
axes[0].set_title('Original')

# MinMax Scaled
sns.histplot(
    data_minmax['sepal length (cm)'],
    kde=True,
    ax=axes[1],
    color='blue',
    alpha=0.5
)
axes[1].set_title('MinMax Scaled')

# Standard Scaled
sns.histplot(
    data_standard['sepal length (cm)'],
    kde=True,
    ax=axes[2],
    color='orange',
    alpha=0.5
)
axes[2].set_title('Standard Scaled (Media=0, Std=1)')

plt.tight_layout()
plt.show()
```

### 3. Crea una gráfica de dispersión entre 'sepal length (cm)' y 'sepal width (cm)' con los datos escalados por StandardScaler.

```python
plt.figure(figsize=(8, 6))
plt.scatter(
    data_standard['sepal length (cm)'],
    data_standard['sepal width (cm)'],
    alpha=0.7,
    color='purple',
    edgecolor='black',
)
plt.title(
    'Dispersión con StandardScaler (sepal length vs sepal width)',
    fontsize=12,
)
plt.xlabel('sepal length (cm) [Estandarizado]')
plt.ylabel('sepal width (cm) [Estandarizado]')
plt.axhline(0, color='gray', linestyle='--', linewidth=0.8)
plt.axvline(0, color='gray', linestyle='--', linewidth=0.8)
plt.grid(True, alpha=0.3)
plt.show()
```

### 4. ¿Qué técnica de escalado crees que es más adecuada para este conjunto de datos? ¿Por qué?

```python
Para el conjunto de datos Iris, StandardScaler es la técnica más adecuada porque los modelos típicos para este dataset (como KNN, SVM o K-Means) calculan distancias euclidianas; estandarizar con media 0 y varianza 1 evita que una variable con mayor dispersión sesgue el cálculo. Al no tener outliers extremos, StandardScaler representa fielmente la distribución natural de las características de las flores. (Se usaría MinMaxScaler únicamente si el algoritmo destino exigiera rangos acotados estrictos entre 0 y 1, como ciertas redes neuronales).
```
